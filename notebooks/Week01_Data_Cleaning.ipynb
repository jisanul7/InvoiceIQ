## 5. Date Column Validation

We will inspect the invoice, due, and payment dates to identify invalid date values.

```python
date_columns = ["invoice_date", "due_date", "payment_date"]

for column in date_columns:
    train_sample[column] = pd.to_datetime(
        train_sample[column],
        errors="coerce"
    )

    print(f"\n{column}")
    print("Invalid or missing dates:", train_sample[column].isna().sum())
```

```python
# --- CELL 1: Temporal & Value Validations ---
invalid_due = train_sample['due_date'] < train_sample['invoice_date']
invalid_pay = (train_sample['payment_status'] == 'Paid') & (train_sample['payment_date'] < train_sample['invoice_date'])
invalid_amt = train_sample['total_amount'] <= 0

print(f"Due date before invoice date: {invalid_due.sum()}")
print(f"Payment date before invoice date: {invalid_pay.sum()}")
print(f"Non-positive total amounts: {invalid_amt.sum()}")

# Filter out impossible records if any exist
cleaned_df = train_sample[~(invalid_due | invalid_amt)].copy()
```

```python
# --- 1. Standardization ---
text_cols = ['currency', 'customer_country', 'payment_status', 'payment_method', 'product_category']
for col in text_cols:
    if col in cleaned_df.columns:
        cleaned_df[col] = cleaned_df[col].astype(str).str.strip()

cleaned_df['currency'] = cleaned_df['currency'].str.upper()

# --- 2. Base Target Columns (Leakage Prevention) ---
# 1 = Unpaid/Pending/Open, 0 = Paid/Closed
cleaned_df['open_paid_flag'] = cleaned_df['payment_status'].str.lower().isin(['pending', 'open']).astype(int)

# Settlement metrics computed ONLY for closed/paid invoices
paid_mask = cleaned_df['open_paid_flag'] == 0

cleaned_df['days_to_pay'] = np.nan
cleaned_df.loc[paid_mask, 'days_to_pay'] = (
    cleaned_df.loc[paid_mask, 'payment_date'] - cleaned_df.loc[paid_mask, 'invoice_date']
).dt.days

cleaned_df['delay'] = np.nan
cleaned_df.loc[paid_mask, 'delay'] = (
    cleaned_df.loc[paid_mask, 'payment_date'] - cleaned_df.loc[paid_mask, 'due_date']
).dt.days

cleaned_df['paid_late_flag'] = np.nan
cleaned_df.loc[paid_mask, 'paid_late_flag'] = (
    cleaned_df.loc[paid_mask, 'delay'] > 0
).astype(int)

# --- 3. Export Cleaned Dataset & Cleaning Audit Log ---
from pathlib import Path

output_dir = Path("../data/processed")
output_dir.mkdir(parents=True, exist_ok=True)

# Save Parquet & CSV files
cleaned_df.to_parquet(output_dir / "cleaned_invoices.parquet", index=False)
cleaned_df.to_csv(output_dir / "cleaned_invoices.csv", index=False)

# Audit log summary
cleaning_log_df = pd.DataFrame([
    {
        "Issue": "Date Parsing",
        "Action Taken": "Converted to datetime64[ns]",
        "Reason": "Enable date arithmetic and interval features",
        "Rows Affected": len(cleaned_df)
    },
    {
        "Issue": "Due Date < Invoice Date",
        "Action Taken": "Audited and verified",
        "Reason": "Chronological coherence of payment terms",
        "Rows Affected": int(invalid_due.sum())
    },
    {
        "Issue": "Payment Date < Invoice Date",
        "Action Taken": "Audited and verified",
        "Reason": "Payment cannot precede invoice date",
        "Rows Affected": int(invalid_pay.sum())
    },
    {
        "Issue": "Non-positive Total Amount",
        "Action Taken": "Audited and verified",
        "Reason": "Transactions must reflect positive gross receivables",
        "Rows Affected": int(invalid_amt.sum())
    },
    {
        "Issue": "Target Leakage on Open Records",
        "Action Taken": "Set delay & days_to_pay to NaN for pending invoices",
        "Reason": "Prevent ground truth leakage into unsettled records",
        "Rows Affected": int((cleaned_df["open_paid_flag"] == 1).sum())
    }
])

cleaning_log_df.to_csv(output_dir / "cleaning_log.csv", index=False)
print("Files exported to ../data/processed successfully!")
display(cleaning_log_df)
display(cleaned_df[['invoice_number', 'payment_status', 'open_paid_flag', 'days_to_pay', 'delay', 'paid_late_flag']].head())
```
